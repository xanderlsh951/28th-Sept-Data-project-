This dataset provides cross-sectional lifestyle telemetry for 8,500 individuals, mapping their evening tech habits, caffeine intake, and demographic chronotypes against physiological sleep metrics and resulting fatigue scores.

````python
primary_bedtime_app = sleep_copy_df['primary_bedtime_app'].nunique()
````

````python
sleep_copy_df['bedtime_phone_minutes'] = pd.to_numeric(sleep_copy_df['bedtime_phone_minutes'], errors='coerce')
````

````python
sleep_copy_df['primary_bedtime_app'].value_counts()
````

Find out if there's any missing rows

````python

import numpy as np
````

````python
#this is finding percentage of missing rows in each column
missing_ratio = sleep_copy_df.isna().mean() * 100
missing_ratio
````

According to above, there's no missing rows.

````python
#this is counting all non-null/valid entries down each column (which defaults to axis=0). Because .isna() replaces values with True and False (neither of which are NaN), .count() treats every single one of them as valid and returns the total row count (8,500) for every column.
missing_num = sleep_copy_df.isna().count()
missing_num
````

````python
heavy_missing_cols = missing_ratio[missing_ratio > 30]
heavy_missing_cols
````

````python
# This will correctly return the number of missing rows per column
missing_num2 = sleep_copy_df.isna().sum()
print(missing_num2)
````
