## Function 3 - np.expand_dims()

Expand the shape of an array.

Insert a new axis that will appear at the axis position in the expanded array shape.

```python
# Example 1 - working
arr1=np.array([
    [[11, 12, 13],
     [13, 14, 15]],
    [[15, 16, 17],
     [17, 18, 19.5]]])
print("Original array:\n", arr1)
print("Original shape:", arr1.shape)

arr_expanded=np.expand_dims(arr1,axis=0)

print("Expanded array: \n", arr_expanded)
print("Expanded shape:", arr_expanded.shape)
```

```python
arr = np.array([1, 2, 3])
print("Original array:", arr)
print("Original shape:", arr.shape)

# Expand dimensions at axis 0 (beginning)
expanded_arr_0 = np.expand_dims(arr, axis=0)
print("expanded shape:", expanded_arr_0.shape)

# add a new row to the expanded array

new_row = np.array([4,5,6])
# Fix: Expand new_row to have 2 dimensions to match expanded_arr_0
new_row_expanded = np.expand_dims(new_row, axis=0) # Or simply new_row_expanded = np.array([[4,5,6]])
final_row = np.append(expanded_arr_0, new_row_expanded, axis=0)

print("\nExpanded at axis 0:", expanded_arr_0)
print("\nafter adding new row:\n", final_row)
print("Shape after expanding at axis 0:", final_row.shape)
```

## Function 3 - np.clip()

function limits the values in an array to be within a specified minimum and maximum range.

```python
# Example 1 - clip values so that the maximum and minimum can only take certain values

arr = np.array([1,2,3,4,5,6,7])

# between 3 and 5

between_arr = np.clip(arr,3,5)
print("the resulting array:",between_arr)
```

Explanation about example

```python
# Example 2 - since between 1-7, the minimum and maximum is 1 and 7, so the values 1,2,6,7 will be clipped, making the 1,2 into 3s and 6,7 into 5s
```

## Function 4 - np.meshgrid()

creates coordinate matrices from coordinate vectors.

```python
# Example 1 - given two arrays of same shape, mesh them into points (pairing of x and y coordinates)
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

A,B = np.meshgrid(a,b)

print("X Grid (X-coordinates for every point):\n", A)

print("Y Grid (Y-coordinates for every point):\n", B)
```

```python
# Example 2 - what if using 2D arrays
a = np.array([[1, 2, 3],
              [4, 5, 6]])
b = np.array([[7, 8, 9],
             [10, 11, 12]])

A,B = np.meshgrid(a,b)

print("X Grid (X-coordinates for every point):\n", A)

print("Y Grid (Y-coordinates for every point):\n", B)

# as you can see, it can still do 2D arrays, it will just consider every a,b and map them into 36 points
```

Explanation about example

```python
# Example 3 - what about using different shapes?
a = np.array([1, 2, 3])
b = np.array([4, 5])

A,B = np.meshgrid(a,b)

print("X Grid (X-coordinates for every point):\n", A)

print("Y Grid (Y-coordinates for every point):\n", B)

# as you can see from the result, instead of 9 points, it only has 6 points
```
